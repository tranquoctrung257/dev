```python
class Student:
    def __init__(self): # luôn luôn có hàm init
        self.name = None
        self.age = None
        self.math_score = None
        self.literature_score = None
        self.teacher = "dung lai"
        # teacher = "dung lai" # nếu sử dụng như thế này thì được hiểu là cái biến này chỉ có tác dụng trong hàm này thôi.

        
    def average_score(self):
    # def average_score(self): nếu không có hàm self ở đây thì hàm này là hàm bộ trợ thôi
        ave_score = (self.math_score + self.literature_score)/2
        # ave_score = (math_score + literature_score)/2 nếu như này sẽ báo lỗi
        return ave_score

    def print_info(self):
        return (f"Student name {self.name} study with {self.teacher}")
S1 = Student()
S1.name = "David"
S1.age = 21
S1.math_score = 9
S1.literature_score = 7

print(S1.teacher)
# nếu muốn đổi teacher
S1.teacher = "Jane"
print(S1.teacher)

print(S1.average_score())
S2 = Student()
S2.name = "Bob"
S2.age = 19
S2.math_score = 5
S2.literature_score = 6
print(S2.average_score())

# nếu mà gọi hàm trong class thì phải có ()
print(S1.print_info())
```

```python
# để đổi lại ngắn gọn hơn thì thêm tham số vào hàm

class Student:
    def __init__(self,name,age,math_score,literature_score): 
        self.name = name
        self.age = age
        self.math_score = math_score
        self.literature_score = literature_score
        self.teacher = "dung lai"
      

        
    def average_score(self):
        ave_score = (self.math_score + self.literature_score)/2
        return ave_score

    def print_info(self):
        return (f"Student name {self.name} study with {self.teacher} have average score {self.average_score()}")



students = []
S1 = Student("David",21,9,7)
S2 = Student("Bob",19,5,6)
S3 = Student("Jenny",29,10,6)

# ví dụ sửa lại tên giáo viên học sinh 3
S3.teacher = "Jane"
students.append(S1)
students.append(S2)
students.append(S3)


for i in range(len(students)):
    print(students[i].print_info())
```
