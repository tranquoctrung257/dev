```python
# 23 param cái cho vào hàm đó.
# max_number trong này gọi là param 
def print_number(max_number):
    for i in range(max_number):
        print(i)

print_number(5)
```

```python
# 24

# ví dụ nhập số 12,14 in ra các số từ 12 đến 14 

def print_number(min_number,max_number,distance):
    for i in range(min_number,max_number+1,distance):
        print(i)

print_number(1,21,2)
```
